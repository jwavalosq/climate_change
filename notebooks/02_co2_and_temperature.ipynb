```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from pathlib import Path
ROOT = Path('..').resolve()
PROC = ROOT / 'data' / 'processed'
CH = ROOT / 'charts'
CH.mkdir(exist_ok=True)
plt.rcParams.update({'figure.dpi': 150})
print('Paths ready:', PROC.exists(), CH.exists())
```

# 2. Carbon dioxide and temperature together

Plain words first. CO2 means carbon dioxide, a gas in the air. ppm means parts per million. If air had one million parts, 421 ppm means 421 of them are CO2. Mauna Loa is a NOAA measuring station in Hawaii, far from most factories and cities, so it gives a clean background reading.

Correlation is a number from minus 1 to plus 1 that says whether two series move together. A value near plus 1 means when one goes up, the other usually goes up too. Correlation alone does not prove that one thing caused the other. This notebook states the correlation plainly and does not claim it proves a sole cause.

Sources: NOAA Global Monitoring Laboratory, Mauna Loa annual means, 1959 to 2025. NASA GISTEMP global anomalies, 1880 to 2023. The joined table covers 1959 to 2023.

```python
c = pd.read_csv(PROC / 'co2_annual.csv')
print(c.head())
print(c.tail())
print('CO2 in 1959:', c[c['year']==1959]['mean_ppm'].iloc[0], 'ppm')
print('CO2 in 2023:', c[c['year']==2023]['mean_ppm'].iloc[0], 'ppm')
print('CO2 in 2025:', c[c['year']==2025]['mean_ppm'].iloc[0], 'ppm')
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(c['year'], c['mean_ppm'], color='#9467bd', linewidth=2)
ax.set_title('Carbon dioxide at Mauna Loa, annual mean, 1959 to 2025')
ax.set_xlabel('Year')
ax.set_ylabel('Parts per million (ppm)')
fig.savefig(CH / '05_co2_annual.png', bbox_inches='tight')
plt.show()
```

```python
j = pd.read_csv(PROC / 'co2_temp_annual.csv')
corr = j['mean_ppm'].corr(j['annual_anomaly_c'])
print('Years joined:', len(j), j['year'].min(), 'to', j['year'].max())
print(f'Correlation between annual CO2 and annual temperature anomaly: {corr:.3f}')
print('Reading: a strong positive correlation. It shows the two rose together.')
print('It does not, on its own, prove that CO2 was the sole cause.')
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(j['mean_ppm'], j['annual_anomaly_c'], s=20, color='#1f77b4')
ax.set_title('Each dot is one year: higher CO2, higher temperature anomaly')
ax.set_xlabel('CO2 in ppm, Mauna Loa annual mean')
ax.set_ylabel('Temperature anomaly in degrees Celsius')
fig.savefig(CH / '07_co2_vs_temperature_scatter.png', bbox_inches='tight')
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(10, 5))
ax1.plot(j['year'], j['mean_ppm'], color='#9467bd', linewidth=2)
ax1.set_xlabel('Year')
ax1.set_ylabel('CO2 in ppm', color='#9467bd')
ax2 = ax1.twinx()
ax2.plot(j['year'], j['annual_anomaly_c'], color='#d62728', linewidth=1.5)
ax2.set_ylabel('Temperature anomaly in degrees Celsius', color='#d62728')
ax1.set_title('CO2 and global temperature rose together, 1959 to 2023')
fig.savefig(CH / '06_co2_and_temperature_dual.png', bbox_inches='tight')
plt.show()
print('CO2 rise 1959 to 2025:', round(c[c['year']==2025]['mean_ppm'].iloc[0] - c[c['year']==1959]['mean_ppm'].iloc[0], 2), 'ppm')
```

## What this shows, and what it does not show

From 1959 to 2023 the correlation is 0.964, a strong positive value. CO2 rose from 315.98 ppm in 1959 to 421.08 ppm in 2023, and to 427.35 ppm in 2025. Temperature anomalies rose over the same span. That joint rise is a fact from these two datasets. It is not, by itself, a full causal proof. A full attribution case uses physics, many datasets, and models, which are outside this project's scope. This notebook stays inside what these two series can honestly show.
