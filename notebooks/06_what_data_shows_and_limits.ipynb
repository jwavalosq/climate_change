# 6. What the data shows, and what it does not show

This final notebook is a honesty check. It repeats the headline numbers from the earlier notebooks, recomputes them from the processed tables, and lists the limits in plain words. If a claim cannot be recomputed here, it does not belong in the story.

No new data is introduced in this notebook. That is on purpose: the point is to audit, not to add.

```python
g = pd.read_csv(PROC / 'global_annual.csv')
d = pd.read_csv(PROC / 'decade_means.csv')
c = pd.read_csv(PROC / 'co2_annual.csv')
j = pd.read_csv(PROC / 'co2_temp_annual.csv')
ev = pd.read_csv(PROC / 'disasters_events.csv')
z = pd.read_csv(PROC / 'zonal_annual.csv')
checks = []
checks.append(('Global years in repo file', len(g), 144))
checks.append(('Disaster events 1980-2024', len(ev), 403))
checks.append(('CO2 years 1959-2025', len(c), 67))
checks.append(('Joined CO2 and temperature years', len(j), 65))
for label, got, want in checks:
    print(label, ':', got, '| expected:', want, '| match:', got == want)
    assert got == want
print('Correlation CO2 vs temperature, 1959-2023:', round(j['mean_ppm'].corr(j['annual_anomaly_c']), 3))
print('Total disaster cost, billion USD CPI adjusted:', round(ev['cost_cpi_adjusted_billion_usd'].sum(), 1))
print('Zonal global 2024:', z[z['year']==2024]['global_c'].iloc[0], '| 2025:', z[z['year']==2025]['global_c'].iloc[0])
print('Note: 2024 and 2025 zonal values come from the current NASA file, not the repo file, which ends in 2023.')
```

## What the data shows

1. Global temperature anomalies rose from a 1880s average of minus 0.209 degrees to a 2010s average of plus 0.806 degrees, on the 1951 to 1980 baseline, in the repo GISTEMP file.
2. CO2 at Mauna Loa rose from 315.98 ppm in 1959 to 427.35 ppm in 2025, with a 0.964 correlation to annual temperature anomalies over 1959 to 2023.
3. The far north warmed about 3.1 times as much as the globe, comparing 2014-2023 with 1880-1889.
4. United States billion-dollar disasters rose from 33 events in the 1980s to 131 in the 2010s, with 115 more in 2020 to 2024, for a 1980 to 2024 total of 403 events and about 2,917.6 billion dollars CPI adjusted.

## What the data does not show

1. It does not show the temperature of any single town or day. These are global and zonal averages.
2. Correlation between CO2 and temperature does not, alone, prove a sole cause. Other factors, feedbacks, and natural variation also matter, and this project does not model them.
3. The disaster file covers the United States only, counts only events over 1 billion dollars, and ends in 2024. It cannot prove a global trend, and it cannot separate climate change from more property at risk or better reporting.
4. Costs are estimates, CPI adjusted, with uncertainty. NOAA itself publishes confidence bounds in its time series file. This project uses point estimates and says so.
5. Early temperature records are thinner than modern ones, especially in the far north and far south. That is a reason to trust decade averages more than any single early year.
6. Disputed figures are not settled here. Where different sources give different disaster costs, for example insurer estimates versus NOAA estimates, this project uses NOAA and labels it as NOAA, so a reader can compare.

## Uncertainty, in one paragraph

Every number here comes with a source, a year range, and a unit. Where a range is partial, it is labelled partial. Where a dataset stopped, 2024 for disasters and 2023 for the repo temperature file, the story stops there for that dataset, instead of filling the gap with a guess. That restraint is part of the result.
